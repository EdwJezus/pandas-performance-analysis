```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["media_final"], bins=5, color="steelblue", edgecolor="black")
plt.xlabel("Média Final")
plt.ylabel("Quantidade de Alunos")
plt.title("Distribuição das Médias Finais")
plt.tight_layout()
plt.show()
```

```python
df["media_final"] = df[
    ["nota_p1", "nota_p2", "nota_p3"]
].mean(axis=1)

cores = df["aprovado"].map({
    "Sim": "green",
    "Não": "red"
})

plt.figure(figsize=(8,5))

plt.scatter(
    df["faltas"],
    df["media_final"],
    c=cores,
    s=80,
    edgecolors="gray"
)

plt.title("Faltas vs Média Final")
plt.xlabel("Número de Faltas")
plt.ylabel("Média Final")

plt.axhline(
    7.0,
    color="blue",
    linestyle="--",
    label="Limite Aprovação (7.0)"
)

plt.legend()
plt.tight_layout()
plt.show()
```
